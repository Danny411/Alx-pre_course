```python
# prompt: Generate meaningful features such as moving averages, technical indicators, and volatility measures for NVDA.

import pandas as pd
import numpy as np

# Ensure we clean up multi-index columns if present (from yfinance v0.2+)
if isinstance(data_nvda.columns, pd.MultiIndex):
    data_nvda.columns = data_nvda.columns.get_level_values(0)

# 1. Moving Averages
data_nvda['MA_5'] = data_nvda['Close'].rolling(window=5).mean()
data_nvda['MA_10'] = data_nvda['Close'].rolling(window=10).mean()

# 2. Volatility Measure (10-day rolling standard deviation of daily returns)
daily_returns = data_nvda['Close'].pct_change()
data_nvda['Volatility'] = daily_returns.rolling(window=10).std()

# 3. Technical Indicator: RSI (Relative Strength Index - 14 days)
delta = data_nvda['Close'].diff()
gain = delta.clip(lower=0)
loss = -1 * delta.clip(upper=0)
ema_gain = gain.ewm(com=13, adjust=False).mean()
ema_loss = loss.ewm(com=13, adjust=False).mean()
rs = ema_gain / ema_loss
data_nvda['RSI_14'] = 100 - (100 / (1 + rs))

# Print the tail of the data frame to verify features
print(data_nvda[['Close', 'MA_5', 'MA_10', 'Volatility', 'RSI_14']].tail())
```

```python
# prompt: Adopt the first four kinds of features and train a regression model to predict tomorrow's stock price. Split your trainining and testing data by a 70/30 ratio and report your performance metric

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Create a copy of the NVDA dataset to work with
df_model = data_nvda.copy()

# Define the target variable: Tomorrow's Close Price (shifted back by 1 day)
df_model['Target'] = df_model['Close'].shift(-1)

# Define feature columns based on the first four categories:
# 1. Historical Prices (Open, High, Low, Close)
# 2. Volume
# 3. Moving Averages (MA_5, MA_10)
# 4. Technical Indicators (RSI_14)
feature_cols = ['Open', 'High', 'Low', 'Close', 'Volume', 'MA_5', 'MA_10', 'RSI_14']

# Drop any rows with NaN values resulting from rolling windows and target shifting
df_clean = df_model[feature_cols + ['Target']].dropna()

X = df_clean[feature_cols]
y = df_clean['Target']

# Train/Test Split by 70/30 ratio (shuffle=False is standard for time-series datasets)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, shuffle=False)

# Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate performance metrics
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"--- Linear Regression Model Performance (NVDA) ---")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R²) Score: {r2:.4f}")
```

```python

# prompt: Train a regression model using price-based, momentum,
# trend, and volatility features to predict tomorrow's NVDA stock price.
# Apply a 70/30 training-testing split and report performance metrics.

!pip install -q yfinance scikit-learn

import yfinance as yf
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Load historical stock data
data = yf.download("NVDA", start="2021-01-01", end="2026-01-01",
                   auto_adjust=True, progress=False)

if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

# 1. Price-based features
data["Price_Range"] = data["High"] - data["Low"]
data["Open_Close_Diff"] = data["Close"] - data["Open"]

# 2. Momentum features
data["ROC_10"] = data["Close"].pct_change(10) * 100
low = data["Low"].rolling(14).min()
high = data["High"].rolling(14).max()
data["Stochastic_K"] = ((data["Close"] - low) / (high - low)) * 100

# 3. Trend features
data["SMA_20"] = data["Close"].rolling(20).mean()
data["SMA_50"] = data["Close"].rolling(50).mean()
data["EMA_12"] = data["Close"].ewm(span=12).mean()
data["EMA_26"] = data["Close"].ewm(span=26).mean()

# 4. Volatility features
returns = data["Close"].pct_change()
data["Volatility_10"] = returns.rolling(10).std()
data["Volatility_30"] = returns.rolling(30).std()
data["Volatility_60"] = returns.rolling(60).std()
data["ATR_14"] = pd.concat([
    data["High"] - data["Low"],
    (data["High"] - data["Close"].shift()).abs(),
    (data["Low"] - data["Close"].shift()).abs()
], axis=1).max(axis=1).rolling(14).mean()

# Define features and prediction target
features = [
    "Price_Range", "Open_Close_Diff",
    "ROC_10", "Stochastic_K",
    "SMA_20", "SMA_50", "EMA_12", "EMA_26",
    "Volatility_10", "Volatility_30",
    "Volatility_60", "ATR_14"
]

data["Target"] = data["Close"].shift(-1)
dataset = data[features + ["Target"]].replace(
    [np.inf, -np.inf], np.nan
).dropna()

X = dataset[features]
y = dataset["Target"]

# Split dataset chronologically (70/30)
split = int(len(dataset) * 0.70)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

# Train regression model
model = RandomForestRegressor(
    n_estimators=200,
    max_depth=10,
    min_samples_split=5,
    random_state=42,
    n_jobs=-1
)
model.fit(X_train, y_train)

# Generate predictions and evaluate
predictions = model.predict(X_test)
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

# Report performance in the required format
print("Regression model performance:")
print(f"- Mean Squared Error (MSE): {mse:.2f}")
print(f"- R-squared (R²) Score: {r2:.3f}")

print("\nPerformance metric report:")
if r2 < 0:
    print(
        "These metrics indicate that the model is not performing well. "
        "The negative R² score suggests that the model performs worse "
        "than predicting the average target value. The selected features "
        "and model may require further optimization to improve prediction "
        "accuracy and generalization."
    )
else:
    print(
        f"The model achieved an MSE of {mse:.2f} and an R² score of "
        f"{r2:.3f}. These results indicate that the model explains "
        f"{max(0, r2) * 100:.2f}% of the variation in the target variable. "
        "Further evaluation and optimization may improve prediction accuracy."
    )
```

```python

# prompt: Create a binary target variable (1 if NVDA closing price
# increases tomorrow, 0 otherwise). Train a Random Forest Classifier
# using a 70/30 train-test split. Report accuracy, confusion matrix,
# classification report, and feature importance.

!pip install -q yfinance scikit-learn

import yfinance as yf
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

# Load historical NVDA stock data
data = yf.download(
    "NVDA",
    start="2021-01-01",
    end="2026-01-01",
    auto_adjust=True,
    progress=False
)

if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

# Generate moving average features
data["MA_5"] = data["Close"].rolling(5).mean()
data["MA_10"] = data["Close"].rolling(10).mean()
data["MA_50"] = data["Close"].rolling(50).mean()

# Create binary target variable
data["Target"] = (
    data["Close"].shift(-1) > data["Close"]
).astype(int)

# Define features
features = [
    "Open", "High", "Low", "Close",
    "Volume", "MA_5", "MA_10", "MA_50"
]

# Prepare dataset
dataset = data[features + ["Target"]].replace(
    [np.inf, -np.inf], np.nan
).dropna()

X = dataset[features]
y = dataset["Target"]

# Split data chronologically (70/30)
split = int(len(dataset) * 0.70)

X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

# Train Random Forest Classifier
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

# Generate predictions
predictions = model.predict(X_test)

# Calculate performance metrics
accuracy = accuracy_score(y_test, predictions)
cm = confusion_matrix(y_test, predictions, labels=[0, 1])
report = classification_report(
    y_test, predictions, zero_division=0
)

# Calculate feature importance
importance = pd.Series(
    model.feature_importances_,
    index=features
).sort_values(ascending=False)

# Display results in required format
print(f"Accuracy: {accuracy * 100:.2f}%")
print(f"\nConfusion Matrix: {cm.tolist()}")
print("\nClassification Report:")
print(report)

print("\nFeature Importance:")
for feature, score in importance.items():
    print(f"{feature} {score:.4f}", end=" ")
print()
```

```python

# prompt: Use the Random Forest Classifier to predict NVDA
# stock price increase/decrease from 2023-01-01 to 2024-01-01.
# Train on historical data before 2023 and report performance metrics.

!pip install -q yfinance scikit-learn

import yfinance as yf
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

# Download historical NVDA data
data = yf.download(
    "NVDA",
    start="2021-01-01",
    end="2024-01-03",
    auto_adjust=True,
    progress=False
)

if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

# Generate moving average features
data["MA_5"] = data["Close"].rolling(5).mean()
data["MA_10"] = data["Close"].rolling(10).mean()
data["MA_50"] = data["Close"].rolling(50).mean()

# Create binary target: 1 = price increases, 0 = decreases
next_close = data["Close"].shift(-1)
data["Target"] = (
    (next_close > data["Close"])
    .where(next_close.notna())
    .astype("float")
)

# Define features
features = [
    "Open", "High", "Low", "Close",
    "Volume", "MA_5", "MA_10", "MA_50"
]

# Prepare dataset
dataset = data[features + ["Target"]].replace(
    [np.inf, -np.inf], np.nan
).dropna()

# Separate training and testing periods
train_data = dataset.loc[dataset.index < "2023-01-01"]
test_data = dataset.loc[
    (dataset.index >= "2023-01-01") &
    (dataset.index < "2024-01-01")
]

X_train = train_data[features]
y_train = train_data["Target"].astype(int)

X_test = test_data[features]
y_test = test_data["Target"].astype(int)

# Train Random Forest Classifier
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

# Predict stock price direction
predictions = model.predict(X_test)

# Calculate performance metrics
accuracy = accuracy_score(y_test, predictions)
cm = confusion_matrix(y_test, predictions, labels=[0, 1])
report = classification_report(
    y_test, predictions, zero_division=0
)

# Display performance results
print(f"Accuracy: {accuracy * 100:.2f}%")
print(f"\nConfusion Matrix: {cm.tolist()}")
print("\nClassification Report:")
print(report)
```

```python

# prompt: Evaluate the NVDA Random Forest model and simulate
# a confidence-based paper-trading strategy.

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix
)

# Generate prediction probabilities
probabilities = model.predict_proba(X_test)[:, 1]

# Confidence-based signals
threshold = 0.60

signals = (probabilities >= threshold).astype(int)

# Classification performance
print("MODEL PERFORMANCE")
print(f"Accuracy: {accuracy_score(y_test, model.predict(X_test)):.2%}")
print(f"Balanced Accuracy: {balanced_accuracy_score(y_test, model.predict(X_test)):.2%}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, model.predict(X_test)))

print("\nClassification Report:")
print(classification_report(y_test, model.predict(X_test), zero_division=0))

# =====================================
# CONFIDENCE-BASED PAPER TRADING
# =====================================

# Signal is generated after today's close.
# Position is entered at the next session's open
# and exited at that session's close.

next_open = data["Open"].shift(-1).reindex(X_test.index)
next_close = data["Close"].shift(-1).reindex(X_test.index)

# Calculate next-session intraday returns
market_returns = (next_close / next_open) - 1

# Only take bullish positions above the threshold
positions = pd.Series(signals, index=X_test.index)

# Assume 0.1% round-trip transaction cost
transaction_cost = 0.001

strategy_returns = (
    positions * market_returns -
    positions * transaction_cost
)

strategy_returns = strategy_returns.dropna()

# Calculate cumulative returns
cumulative_returns = (
    (1 + strategy_returns).prod() - 1
) * 100

# Calculate market benchmark
benchmark_returns = market_returns.loc[
    strategy_returns.index
].dropna()

benchmark_performance = (
    (1 + benchmark_returns).prod() - 1
) * 100

# Calculate maximum drawdown
equity = (1 + strategy_returns).cumprod()
drawdown = equity / equity.cummax() - 1
max_drawdown = drawdown.min() * 100

# Display results
print("\nPAPER TRADING PERFORMANCE")
print(f"Confidence Threshold: {threshold:.0%}")
print(f"Total Trades: {int(positions.sum())}")
print(f"Strategy Return: {cumulative_returns:.2f}%")
print(f"Benchmark Return: {benchmark_performance:.2f}%")
print(f"Maximum Drawdown: {max_drawdown:.2f}%")
```
